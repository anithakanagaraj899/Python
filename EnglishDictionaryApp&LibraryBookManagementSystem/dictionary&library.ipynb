# English Dictionary App

This project uses the skills learnt from week 1 to week 6. The aim of this project is to create an English Dictionary app that returns definitions of English words

```python
import pickle 
import os

def saveDictionary(tmp_dict):
    try:
       
        with open('words.txt', 'wb') as dictionary:
            pickle.dump(tmp_dict, dictionary)
        
        print("SUCCESS: writing to dictionary words.txt")
    except Exception as e:
        print(f"ERROR: {e}")


def getDictionary():
    try:
        with open('words.txt', 'rb') as dictionary:
            tmp_dict = pickle.load(dictionary)
        print("SUCCESS: reading from dictionary words.txt")
        return tmp_dict
    except Exception as e:
        print(f"ERROR: {e}")



def addNewWord():
    print("*"*30)
    print("NEW WORD ADDER:")
    print("*"*30)
    print("Enter a Word to add")
    word = input(">>")
    print(f"Enter a meaning of {word} to add")
    meaning = input(">>")
    dictionary = getDictionary()
    dictionary[word] = meaning
    saveDictionary(dictionary)

    
    main_menu()


def findMeaning():
    
    print("*"*30)
    print("WORD FINDER:")
    print("*"*30)
    print("Enter a Word to find")
    word = input(">>")
    dictionary = getDictionary()
    if word in dictionary:
        print(f"Found Entry: [{word} : {dictionary[word]}]")
    else:

        print("ERROR: The word doesn't exist!")
        findMeaning()

    
    main_menu()


def updateWord():
    print("*"*30)
    print("WORD UPDATER:")
    print("*"*30)
    print("Enter an old Word to edit")
    word = input(">>")
    dictionary = getDictionary()
    if word in dictionary:
        print(f"Enter a New Meaning to edit [{word} : {dictionary[word]}] to:")
        new_meaning = input(">>")
        dictionary[word] = new_meaning
        saveDictionary(dictionary)
        print(f"SUCCESS: Updated to [{word} : {dictionary[word]}]")
    else:
        print("ERROR: The word doesn't exist! impossible to update!")
        updateWord()

   
    main_menu()


def main_menu():
    if not os.path.exists(os.path.join("words.txt")):
        try:
            with open('words.txt', 'wb') as dictionary:
                pickle.dump({}, dictionary)
            print("SUCCESS: writing to dictionary words.txt")
        except Exception as e:
            print(f"ERROR: {e}")
    print("*"*30)
    print("ENGLISH DICTIONARY APP")
    print("*"*30)
    print("1. Add a new word")
    print("2. Find the meaning")
    print("3. Update a word")
    print("4. Exit")
    print("*"*30)
    n = int(input(">>"))
    if n==1:
        addNewWord()
    elif n==2:
        findMeaning()
    elif n==3:
        updateWord()
    elif n==4:
        exit()
    else:
       
        print("ERROR: Please Enter a digit according to Menu.")
        main_menu()




main_menu() 
```

# Library Book Managment
In this project, a Library Book Management System will be implemented with the front end GUI developed using Tkinter and the backend database implemented in SQLite. It is a single window tkinter GUI application which will implement CRUD( create, read, update, delete) operations on a ‘books’ database, in SQLite. The window will contain text widgets to accept / display book title, author, year of publication, ISBN (optional) and a large text area/ list box to display the records which match the search criteria.

```python

# Libraries
from tkinter import *
import sqlite3
import tkinter.ttk as ttk
import tkinter.messagebox as tkMessageBox
root = Tk()
root.title('A Library Book Managment')
root.configure(bg="orange")
# Methods
def Database():
    global conn, cursor
    conn = sqlite3.connect('books.db')
    cursor = conn.cursor()
    cursor.execute("CREATE TABLE IF NOT EXISTS library (id INTEGER PRIMARY KEY AUTOINCREMENT NOT NULL, booktitle TEXT, author TEXT, year INTEGER, isbn TEXT)") 
def Add():
    if BOOKTITLE.get() == "" or AUTHOR.get() == "" or YEAR.get() == "" or ISBN == "":
        txt_result.config(text = 'Please Enter All the fileds',fg = "red")
    else:
        Database()
        cursor.execute('INSERT INTO library (booktitle,author,year,isbn) VALUES (?,?,?,?)',(str(BOOKTITLE.get()),str(AUTHOR.get()),int(YEAR.get()),str(ISBN.get()),))
        conn.commit()
        BOOKTITLE.set("")
        AUTHOR.set("")
        YEAR.set("")
        ISBN.set("")
        PRICE.set("")
        cursor.close
        conn.close()
        txt_result.config(text = 'Record Inserted',fg = "Green")
def Display():
    tree.delete(*tree.get_children())
    Database()
    cursor.execute('SELECT * FROM library ORDER BY author ASC')
    fetch = cursor.fetchall()
    for data in fetch:
        tree.insert('','end',values = (data[1],data[2],data[3],data[4]))
        cursor.close()
        txt_result.config(text = 'Successfully fetched Data',fg = "white")
def Search():
    if BOOKTITLE.get() =="":
        txt_result.config(text = 'Please Enter the Booktitle ',fg = "Black")
    else :
        tree.delete(*tree.get_children())
        Database()
        cursor.execute('SELECT * FROM library WHERE Booktitle = ?',(BOOKTITLE.get(),))
        fetch = cursor.fetchall()
        for data in fetch:
            tree.insert('','end',values = (data[1],data[2],data[3],data[4]))
            cursor.close()
            txt_result.config(text = 'Successfully fetched Data',fg = "White") 
def Issue():
    result=tkMessageBox.showwarning("Undefined", "this function is not yet defined!")
    conn.commit()
def Delete():
    if BOOKTITLE.get() == "":
        txt_result.config(text = 'Please Enter the Booktitle ',fg = "White ")
    else :
        Database()
        cursor.execute('DELETE FROM library WHERE Booktitle = ?',(BOOKTITLE.get(),))
        conn.commit() 
        BOOKTITLE.set("")
def Exit():
    result = tkMessageBox.askquestion('Do you want to exit ? (y/n)',icon = 'warning')
    if result == "yes":
        root.destroy()
# Variable Declaration
BOOKTITLE = StringVar()
AUTHOR = StringVar()
YEAR = IntVar()
ISBN = StringVar()
PRICE=IntVar() 
# FRAME
Top = Frame(root,width = 1000 , height = 50, bd = 8, relief = 'raise')
Top.pack(side = TOP)
Left = Frame(root,width = 500 , height = 500, bd = 3, relief = 'raise')
Left.pack(side = LEFT)
Right = Frame(root,width = 500, height=500,bd= 3, relief = 'raise')
Right.pack(side = RIGHT)
Forms = Frame(Left,width = 500,height= 450)
Forms.pack(side= TOP)
Buttons = Frame(Left,width = 300 , height = 400, bd = 3, relief = 'raise')
Buttons.pack(side = BOTTOM)
#========================================Labels=================================================================
txt_title = Label(Top, width = 900, font = ('arial',24),text = "...WELCOME TO LIBRARY...")
txt_title.pack()
txt_firstname = Label(Forms,text ='Booktitle', font = ('arial',16), bd = 15)
txt_firstname.grid(row = 0 , stick = "e")
txt_lastname = Label(Forms,text ='Author', font = ('arial',16), bd = 15)
txt_lastname.grid(row = 1 , stick = "e")
txt_lastname = Label(Forms,text ='Year', font = ('arial',16), bd = 15)
txt_lastname.grid(row = 2 , stick = "e")
txt_lastname = Label(Forms,text ='ISBN', font = ('arial',16), bd = 15)
txt_lastname.grid(row = 3 , stick = "e")

txt_result = Label(Buttons)
txt_result.pack(side = TOP)
##========================================Entry Wigets=================================================================
booktitle = Entry(Forms, textvariable = BOOKTITLE, width = 50)
booktitle.grid(row = 0, column = 1)
author = Entry(Forms, textvariable = AUTHOR, width = 50)
author.grid(row = 1, column = 1)
year = Entry(Forms, textvariable = YEAR, width = 50)
year.grid(row = 2, column = 1)
isbn = Entry(Forms, textvariable = ISBN, width = 50)
isbn.grid(row = 3, column = 1)

#========================================Button Widgets=================================================================
btn_display = Button(Buttons, width = 10, text = 'Display', command = Display)
btn_display.pack(side = LEFT)
btn_search = Button(Buttons, width = 10, text = 'Search', command = Search)
btn_search.pack(side = LEFT)
btn_add = Button(Buttons, width = 10, text = 'Add', command = Add)
btn_add.pack(side = LEFT)
btn_issue = Button(Buttons, width = 10, text = 'Issue', command = Issue)
btn_issue.pack(side = LEFT)
btn_delete = Button(Buttons, width = 10, text = 'Delete', command = Delete)
btn_delete.pack(side = LEFT)
btn_exit = Button(Buttons, width = 10, text = 'Exit', command = Exit)
btn_exit.pack(side = LEFT)
##========================================List Widgets=================================================================
scrollbary = Scrollbar(Right, orient = VERTICAL)
scrollbarx = Scrollbar(Right, orient = HORIZONTAL)
tree = ttk.Treeview(Right, columns = ("Booktitle","Author","Year","ISBN"))
scrollbary.config(command = tree.yview)
scrollbary.pack(side = RIGHT,fill =Y)
scrollbarx.config(command = tree.xview)
scrollbarx.pack(side = BOTTOM,fill = X)
tree.heading('Booktitle',text = 'Booktitle',anchor = W)
tree.heading('Author',text = 'Author',anchor = W)
tree.heading('Year',text = 'Year',anchor = W)
tree.heading('ISBN',text = 'ISBN',anchor = W)
tree.column('#0',stretch=NO, minwidth = 0, width =0)
tree.column('#1',stretch=NO, minwidth = 0, width =200)
tree.column('#2',stretch=NO, minwidth = 0, width =120)
tree.column('#3',stretch=NO, minwidth = 0, width =50)
tree.column('#4',stretch=NO, minwidth = 0, width =120)
tree.pack()
# INITIALIZING
if __name__ == '__main__':
    root.mainloop()


        
```
